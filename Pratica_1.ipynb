### Sistema de Recomendação de Viagem

```python
class Destino:
    def __init__(self, nome, clima, tipo, custo):
        self.nome = nome
        self.clima = clima.lower()
        self.tipo = tipo.lower()
        self.custo = custo

    def combina_com(self, clima, tipo, orcamento):
        return (
            self.clima == clima.lower() and
            self.tipo == tipo.lower() and
            self.custo <= orcamento
        )

destinos = [
    Destino("Rio de Janeiro", "Quente", "Cidade", 2000),
    Destino("Gramado", "Frio", "Cidade", 2500),
    Destino("Lençóis Maranhenses", "Quente", "Natureza", 1800),
    Destino("Campos do Jordão", "Frio", "Natureza", 2900)
]
```

```python
print("Boa noite! Vamos encontrar o destino ideal?")

clima = input("Qual tipo de clima você prefere, quente ou frio? ").strip().lower()
tipo = input("Você prefere natureza ou cidade? ").strip().lower()
orcamento = float(input("Qual é o seu orçamento máximo (R$)? "))

for destino in destinos:
    if destino.combina_com(clima=clima, tipo=tipo, orcamento=orcamento):
        print(f"\nSugestão: {destino.nome}")
        print("Boa viagem!")
        break
else:
    print("Sinto muito, mas não encontramos destinos compatíveis com suas preferências!")
```
