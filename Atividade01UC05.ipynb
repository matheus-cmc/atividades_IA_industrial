## ⚠️ 7. Desafios e Limitações
> Quais são as possíveis dificuldades que podem afetar o desempenho do modelo?  
> Exemplos: iluminação, ângulo da câmera, viés dos dados.

*Resposta: Resposta:
1. Variação de ambiente: Mudanças de iluminação podem impedir a detecção do rosto.
2. Oclusão parcial: Uso de máscaras, óculos escuros ou barba excessiva podem dificultar a extração das características faciais.
3. Vieses Algorítmicos: Modelos pré-treinados podem ter desempenhos desiguais para diferentes etnias ou gêneros se não forem calibrados corretamente.
4. Anti-Spoofing: O modelos básico pode ser enganado por uma foto impressa do funcionário (detecção de "Prova de vida")
