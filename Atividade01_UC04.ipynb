```python
# Carrega a imagem onde a detecção será feita.
imagem = cv2.imread('/content/power-rangers-original.webp.jpg')

# Carrega o classificador Haar Cascade pré-treinado para detectar faces frontais.
face_cascade = cv2.CascadeClassifier('haarcascade_frontalface_default.xml')

# Converte a imagem para escala de cinza, pois o detector opera nesta.
imagem_cinza = cv2.cvtColor(imagem, cv2.COLOR_BGR2GRAY)

# Executa a detecção de faces, retornando uma lista de retângulos (x, y, largura, altura).
# scaleFactor=1.3: Define o fator de redução da imagem a cada escala.
# minNeighbors=5: Define quantos vizinhos cada retângulo candidato deve ter para ser considerado uma face.
rostos_detectados = face_cascade.detectMultiScale(imagem_cinza, scaleFactor=1.3, minNeighbors=5)

# Itera sobre cada face (retângulo) que foi detectada.
for (inicio_x, inicio_y, largura_rosto, altura_rosto) in rostos_detectados:

    # Calcula o ponto final (canto inferior direito) do retângulo
    fim_x = inicio_x + largura_rosto
    fim_y = inicio_y + altura_rosto

    # Desenha o retângulo azul ao redor do rosto
    cv2.rectangle(
        img=imagem,        # imagem onde vamos desenhar
        pt1=(inicio_x, inicio_y),   # canto superior esquerdo
        pt2=(fim_x, fim_y),         # canto inferior direito
        color=(255, 0, 0),          # cor azul (em BGR)
        thickness=2                 # espessura da linha
    )

# Converte a imagem com as detecções de BGR para RGB e a exibe.
# Converte a imagem para RGB (senão o matplotlib mostra as cores trocadas)
imagem_rgb = cv2.cvtColor(imagem, cv2.COLOR_BGR2RGB)

# Mostra a imagem com os retângulos
plt.imshow(imagem_rgb)
plt.title("Rostos detectados (em azul)")
plt.axis("off")  # remove os eixos para deixar mais limpo
plt.show()
```
